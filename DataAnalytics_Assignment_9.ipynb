### One-Way ANOVA
We want to test if there is a statistically significant difference in the mean **petal length** (`PetalLengthCm`) among the three species (`Iris-setosa`, `Iris-versicolor`, and `Iris-virginica`).

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Fit the OLS model using PetalLengthCm as the dependent variable and Species as the categorical independent variable
model = ols('PetalLengthCm ~ C(Species)', data=df).fit()

# Perform One-Way ANOVA using the fitted OLS model
anova_table = sm.stats.anova_lm(model, typ=2)

print("ANOVA Table:")
display(anova_table)

# Extract F-statistic and p-value from the ANOVA table
f_val = anova_table['F'].iloc[0]
p_val = anova_table['PR(>F)'].iloc[0]

print(f"\nF-statistic: {f_val:.4f}")
print(f"p-value: {p_val:.4e}")

# Interpret results
alpha = 0.05
if p_val < alpha:
    print("\nResult: Reject the null hypothesis. There is a statistically significant difference in petal length among the three species.")
else:
    print("\nResult: Fail to reject the null hypothesis. There is no statistically significant difference in petal length among the three species.")
```

### Chi-Square Test of Independence
To perform a Chi-Square test, we need two categorical variables. We will create a categorical variable by binning the continuous `PetalLengthCm` column into three categories (`Short`, `Medium`, `Long`), and then test its association with the `Species` column.

### Chi-Square Test: Sepal Width vs Species
We want to test if the sepal width category is independent of the iris species.

```python
# Bin SepalWidthCm into 3 categories
df['SepalWidthCategory'] = pd.qcut(df['SepalWidthCm'], q=3, labels=['Short', 'Medium', 'Long'])

# Create a contingency table
sepal_contingency_table = pd.crosstab(df['Species'], df['SepalWidthCategory'])
print("Contingency Table (Species vs Sepal Width Category):")
display(sepal_contingency_table)
print("\n" + "="*40 + "\n")

# Perform Chi-Square Test of Independence
chi2_sepal, p_val_sepal, dof_sepal, expected_sepal = stats.chi2_contingency(sepal_contingency_table)

print(f"Chi-Square Statistic: {chi2_sepal:.4f}")
print(f"p-value: {p_val_sepal:.4e}")
print(f"Degrees of Freedom: {dof_sepal}")

# Interpret results
alpha = 0.05
if p_val_sepal < alpha:
    print("\nResult: Reject the null hypothesis. There is a statistically significant association between species and sepal width category.")
else:
    print("\nResult: Fail to reject the null hypothesis. There is no significant association between species and sepal width category.")
```
