```python
import numpy as np
import pandas as pd
import scipy.stats as stats
```

```python
# --- 1. DATA LOADING AND EXPLORATION ---
# Set your custom file path here:
file_path = "/content/student_probability_dataset (1) - student_probability_dataset (1) (1).csv"

df = pd.read_csv(file_path)
print(f"Loaded file: {file_path}\n")
print(f"Dataset shape: {df.shape}")
```

```python
# --- 2. EVENT DEFINITION ---
df['A'] = df['Gender'] == 'Male'
df['B'] = df['Result'] == 'Pass'
df['C'] = df['Attendance'] > 75
df['D'] = df['Previous_Score'] > 70
df['E'] = df['Study_Hours'] > 4
```

```python
# --- 3. BASIC PROBABILITY ---
P_A, P_B, P_C, P_D, P_E = df['A'].mean(), df['B'].mean(), df['C'].mean(), df['D'].mean(), df['E'].mean()
print("\n--- 3. Basic Probabilities ---")
print(f"P(Male)       = {P_A:.4f}")
print(f"P(Pass)       = {P_B:.4f}")
print(f"P(Att > 75)   = {P_C:.4f}")
print(f"P(Score > 70) = {P_D:.4f}")
print(f"P(Hours > 4)  = {P_E:.4f}")
```

```python
# --- 4. CONDITIONAL PROBABILITY ---
P_B_given_A = df[df['A']]['B'].mean()
P_B_given_C = df[df['C']]['B'].mean()
P_B_given_D = df[df['D']]['B'].mean()
P_B_given_E = df[df['E']]['B'].mean()
print("\n--- 4. Conditional Probabilities ---")
print(f"P(Pass | Male)       = {P_B_given_A:.4f}")
print(f"P(Pass | Att > 75)   = {P_B_given_C:.4f}")
print(f"P(Pass | Score > 70) = {P_B_given_D:.4f}")
print(f"P(Pass | Hours > 4)  = {P_B_given_E:.4f}")
```

```python
# --- 5. INDEPENDENT EVENTS ---
print("--- 5. Independent Events Analysis ---")

# Helper function to check independence between two boolean series
def check_independence(event1_name, event1, event2_name, event2):
    p_1 = event1.mean()
    p_2 = event2.mean()
    p_joint_empirical = (event1 & event2).mean()
    p_joint_theoretical = p_1 * p_2
    diff = abs(p_joint_empirical - p_joint_theoretical)
    is_independent = diff < 0.01  # Small threshold for numerical/empirical closeness

    print(f"Checking {event1_name} and {event2_name}:")
    print(f"  P({event1_name}) = {p_1:.4f} | P({event2_name}) = {p_2:.4f}")
    print(f"  Empirical P({event1_name} ∩ {event2_name}) = {p_joint_empirical:.4f}")
    print(f"  Theoretical P({event1_name}) * P({event2_name}) = {p_joint_theoretical:.4f}")
    print(f"  Independent? {is_independent} (Difference: {diff:.4f})\n")

# Check some combinations
check_independence("Male (A)", df['A'], "Pass (B)", df['B'])
check_independence("Att > 75 (C)", df['C'], "Pass (B)", df['B'])
check_independence("Score > 70 (D)", df['D'], "Pass (B)", df['B'])
```

```python

# --- 6. UNION, INTERSECTION AND COMPLEMENT ---
P_A_and_B = (df['A'] & df['B']).mean()
P_A_or_B = P_A + P_B - P_A_and_B
P_not_B = 1 - P_B
P_att_le_75 = 1 - P_C
print("\n--- 6. Union, Intersection & Complements ---")
print(f"P(Male AND Pass)      = {P_A_and_B:.4f}")
print(f"P(Male OR Pass)       = {P_A_or_B:.4f}")
print(f"P(Fail)               = {P_not_B:.4f}")
print(f"P(Attendance <= 75)   = {P_att_le_75:.4f}")
```

```python
# --- 7. BAYES' THEOREM ---
P_A_given_B_bayes = (P_B_given_A * P_A) / P_B
P_A_given_B_emp = df[df['B']]['A'].mean()
print("\n--- 7. Bayes' Theorem Verification ---")
print(f"P(Male | Pass) via Bayes' formula : {P_A_given_B_bayes:.4f}")
print(f"P(Male | Pass) direct empirical    : {P_A_given_B_emp:.4f}")
```

```python
# --- 8. DISTRIBUTION ANALYSIS ---
cols = ['Study_Hours', 'Attendance', 'Previous_Score']
desc = df[cols].agg(['mean', 'median', 'std', 'min', 'max']).T
desc['skewness'] = df[cols].skew()
print("\n--- 8. Distribution Statistics ---")
print(desc.round(4))
```

```python
# --- 9. PROBABILITY ASSUMING NORMALITY ---
# Previous Score > 70
mu_ps, std_ps = df['Previous_Score'].mean(), df['Previous_Score'].std()
theo_ps = 1 - stats.norm.cdf(70, mu_ps, std_ps)
emp_ps = (df['Previous_Score'] > 70).mean()

# Attendance < 60
mu_att, std_att = df['Attendance'].mean(), df['Attendance'].std()
theo_att = stats.norm.cdf(60, mu_att, std_att)
emp_att = (df['Attendance'] < 60).mean()

print("\n--- 9. Normality Assumption Analysis ---")
print(f"P(Previous_Score > 70) -> Theoretical: {theo_ps:.4f} | Empirical: {emp_ps:.4f}")
print(f"P(Attendance < 60)     -> Theoretical: {theo_att:.4f} | Empirical: {emp_att:.4f}")
```

```python
# --- 10. CORRELATION ---
print("\n--- 10. Correlation Matrix ---")
print(df[cols].corr().round(4))
```

```python
# --- 11. PERFORMANCE ANALYSIS (Pass Rates) ---
print("\n--- 11. Performance Pass Rates ---")
print("Pass rate by Gender:")
print(df.groupby('Gender')['B'].mean().round(4))

df['Attendance_Group'] = np.where(df['Attendance'] > 75, '> 75%', '<= 75%')
print("\nPass rate by Attendance:")
print(df.groupby('Attendance_Group')['B'].mean().round(4))

df['Score_Group'] = np.where(df['Previous_Score'] > 70, '> 70', '<= 70')
print("\nPass rate by Previous Score:")
print(df.groupby('Score_Group')['B'].mean().round(4))
```
