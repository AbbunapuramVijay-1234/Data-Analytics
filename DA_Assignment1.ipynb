# **Exploring Central Tendency**

```python
mean_values = data_set.mean(numeric_only=True)
print("--- Mean Values ---")
print(mean_values)
```

```python
median_values=data_set.median(numeric_only=True)
print("--- Median Values ---")
print(median_values)
```

```python
# Calculate Mode (the most frequently occurring value)
# .iloc[0] is used because a column can sometimes have multiple modes
mode_values = data_set.mode(numeric_only=True).iloc[0]

print("--- Mode Values ---")
print(mode_values)
```

```python
mean_values = data_set.mean(numeric_only=True)
print("--- Mean (Overall Central Value) ---")
print(mean_values)
```

```python
median_values = data_set.median(numeric_only=True)
print("--- Median (Middle Value) ---")
print(median_values)
```

```python
mode_values = data_set.mode(numeric_only=True).iloc[0]
print("--- Mode (Most Frequently Occurring Value) ---")
print(mode_values)
```

```python
# Select only numerical columns from our dataset
num_data_set = data_set.select_dtypes(include='number')

# 1. Range (Max - Min)
data_range = num_data_set.max() - num_data_set.min()

# 2. Standard Deviation
std_dev = num_data_set.std()

# 3. Variance
variance = num_data_set.var()

# 4. Interquartile Range (IQR = Q3 - Q1)
q1 = num_data_set.quantile(0.25)
q3 = num_data_set.quantile(0.75)
iqr = q3 - q1

# Combine into a clean summary table for spread
spread_table = pd.DataFrame({
    'Range (Max - Min)': data_range,
    'Standard Deviation': std_dev,
    'Variance': variance,
    'IQR (Middle 50% Spread)': iqr
})

print("--- Spread and Variability Analysis ---")
display(spread_table)
```

# **Measures of shape via different visualization charts**

```python
plt.hist(data_set['Price_USD'], bins=10, edgecolor='black')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Histogram of Price')
plt.show()
```

```python
plt.scatter(data_set['Price_USD'], data_set['Size_sqft'])
plt.xlabel('Price')
plt.ylabel('Bedrooms')
plt.title('Scatter Plot of Price vs Size_sqft')
```

```python
plt.boxplot(data_set['Price_USD'],vert=False)
plt.ylabel('Price')
plt.title('Box Plot of Price')
plt.show()
```

```python
h = np.asarray(data_set['Price_USD'])
h = sorted(h)

#use the scipy stats module to fit a normal distirbution with same mean and standard deviation
fit = stats.norm.pdf(h, np.mean(h), np.std(h))

#plot both series on the histogram
plt.plot(h,fit,'-',linewidth = 2,label="Normal distribution with same mean and var")
#plt.hist(h,density=False,bins = 100,label="Actual distribution")
plt.legend()
plt.show()
```
